```python
import requests                 # Library to send HTTP requests (like a browser)
from bs4 import BeautifulSoup   # Library to parse HTML (to read the web page)
import mysql.connector          # Library to talk to MySQL databases
from datetime import datetime   # Library to get the current time
import sys                      # Library to interact with the system (exit, etc.)
import re                       # Library for Regular Expressions (finding patterns in text)

# --- PART 1: AWS CONFIGURATION ---
AWS_HOST = "database-1.cp42ggiaa3k3.ap-south-1.rds.amazonaws.com"
AWS_USER = "admin"
AWS_PASSWORD = "9862768177"
AWS_DB_NAME = "weather_db"

# --- PART 2: DATA SOURCE CONFIGURATION (OPEN-METEO API) ---
# We are switching to Open-Meteo (A Free Weather API).
# Why? Google/Bing HTML changes constantly and blocks bots. 
# APIs are designed specifically for Data Analysts to fetch data reliably.
# Bangalore Coordinates: Latitude 12.9716, Longitude 77.5946
URL = "https://api.open-meteo.com/v1/forecast?latitude=12.9716&longitude=77.5946&current=temperature_2m,relative_humidity_2m&timezone=auto"

def get_weather_data():
    """
    LOGIC: 
    1. Connect to the Open-Meteo API.
    2. Receive the data in JSON format (Structured data).
    3. Extract the Temperature and Humidity directly.
    """
    try:
        print(f"📡 Fetching weather data from Open-Meteo API...")
        
        # Step 1: Send the request
        response = requests.get(URL)
        
        # Debugging: Print status code
        print(f"   Status Code: {response.status_code}")
        
        if response.status_code != 200:
            print("❌ Failed to connect to API.")
            return None

        # Step 2: Parse JSON (Not HTML)
        # APIs return data in a dictionary format, so we don't need BeautifulSoup!
        data = response.json()
        
        # The API returns data like this: 
        # { "current": { "temperature_2m": 28.5, "relative_humidity_2m": 60 } }
        
        current_weather = data.get("current", {})
        
        # Step 3: Extract Data
        city = "Bangalore"
        temperature = current_weather.get("temperature_2m", 0.0)
        humidity = current_weather.get("relative_humidity_2m", 0)
        
        # Get the current time
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        # Return the clean data
        return {
            "city": city,
            "temperature": temperature,
            "humidity": humidity,
            "timestamp": timestamp
        }

    except Exception as e:
        print(f"❌ API Error: {e}")
        return None

def save_to_aws(data):
    """
    LOGIC: Connect to AWS and INSERT the row.
    """
    connection = None
    try:
        print(f"☁️ Connecting to AWS...")
        
        connection = mysql.connector.connect(
            host=AWS_HOST,
            user=AWS_USER,
            password=AWS_PASSWORD,
            database=AWS_DB_NAME,
            port=3306
        )
        
        cursor = connection.cursor()
        
        sql_query = """
        INSERT INTO weather_readings (city, temperature, humidity, timestamp)
        VALUES (%s, %s, %s, %s)
        """
        
        values = (data['city'], data['temperature'], data['humidity'], data['timestamp'])
        
        cursor.execute(sql_query, values)
        connection.commit()
        
        print(f"✅ SUCCESS: Saved {data['city']} data ({data['temperature']}°C, {data['humidity']}%) to AWS!")
        
    except mysql.connector.Error as e:
        print(f"❌ Database Error: {e}")
    finally:
        if connection and connection.is_connected():
            connection.close()

if __name__ == "__main__":
    # 1. First, try to get the data
    weather_data = get_weather_data()
    
    # 2. If data was successfully fetched, THEN try to save it
    if weather_data:
        save_to_aws(weather_data)
    else:
        print("⚠️ Skipped database save due to error.")
```

```python
import requests
import mysql.connector
from datetime import datetime
import time
import sys
import os
import heapq  # DSA: Priority Queue Algorithm
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

# --- PART 1: AWS CONFIGURATION ---
AWS_HOST = os.environ.get("DB_HOST", "database-1.cp42ggiaa3k3.ap-south-1.rds.amazonaws.com")
AWS_USER = os.environ.get("DB_USER", "admin")
AWS_PASSWORD = os.environ.get("DB_PASS", "9862768177")
AWS_DB_NAME = "weather_db"

# --- PART 2: DSA PRIORITY CONFIGURATION ---
# Tier 1 = Every 15 mins (900s)
# Tier 2 = Every 30 mins (1800s)
# Tier 3 & North East = Every 60 mins (3600s)
TIER_INTERVALS = {
    "1": 900,
    "2": 1800,
    "3": 3600,
    "North East": 3600
}

# FULL CITY LIST (60+ Cities)
CITIES = [
    # --- TIER 1 ---
    {"name": "Bangalore", "tier": "1", "lat": 12.9716, "lon": 77.5946},
    {"name": "Delhi",     "tier": "1", "lat": 28.7041, "lon": 77.1025},
    {"name": "Mumbai",    "tier": "1", "lat": 19.0760, "lon": 72.8777},
    {"name": "Chennai",   "tier": "1", "lat": 13.0827, "lon": 80.2707},
    {"name": "Hyderabad", "tier": "1", "lat": 17.3850, "lon": 78.4867},
    {"name": "Kolkata",   "tier": "1", "lat": 22.5726, "lon": 88.3639},
    {"name": "Pune",      "tier": "1", "lat": 18.5204, "lon": 73.8567},
    {"name": "Ahmedabad", "tier": "1", "lat": 23.0225, "lon": 72.5714},

    # --- NORTH EAST ---
    {"name": "Guwahati",  "tier": "North East", "lat": 26.1445, "lon": 91.7362},
    {"name": "Shillong",  "tier": "North East", "lat": 25.5788, "lon": 91.8933},
    {"name": "Agartala",  "tier": "North East", "lat": 23.8315, "lon": 91.2868},
    {"name": "Silchar",   "tier": "North East", "lat": 24.8333, "lon": 92.7789},
    {"name": "Imphal",    "tier": "North East", "lat": 24.8170, "lon": 93.9368},
    {"name": "Aizawl",    "tier": "North East", "lat": 23.7271, "lon": 92.7176},
    {"name": "Kohima",    "tier": "North East", "lat": 25.6701, "lon": 94.1077},
    {"name": "Gangtok",   "tier": "North East", "lat": 27.3314, "lon": 88.6138},
    {"name": "Itanagar",  "tier": "North East", "lat": 27.0844, "lon": 93.6053},
    {"name": "Dimapur",   "tier": "North East", "lat": 25.8629, "lon": 93.7533},
    {"name": "Tezpur",    "tier": "North East", "lat": 26.6528, "lon": 92.7926},
    {"name": "Dibrugarh", "tier": "North East", "lat": 27.4728, "lon": 94.9120},

    # --- TIER 2 ---
    {"name": "Surat", "tier": "2", "lat": 21.1702, "lon": 72.8311},
    {"name": "Lucknow", "tier": "2", "lat": 26.8467, "lon": 80.9462},
    {"name": "Jaipur", "tier": "2", "lat": 26.9124, "lon": 75.7873},
    {"name": "Kanpur", "tier": "2", "lat": 26.4499, "lon": 80.3319},
    {"name": "Nagpur", "tier": "2", "lat": 21.1458, "lon": 79.0882},
    {"name": "Indore", "tier": "2", "lat": 22.7196, "lon": 75.8577},
    {"name": "Thane", "tier": "2", "lat": 19.2183, "lon": 72.9781},
    {"name": "Bhopal", "tier": "2", "lat": 23.2599, "lon": 77.4126},
    {"name": "Visakhapatnam", "tier": "2", "lat": 17.6868, "lon": 83.2185},
    {"name": "Patna", "tier": "2", "lat": 25.5941, "lon": 85.1376},
    {"name": "Vadodara", "tier": "2", "lat": 22.3072, "lon": 73.1812},
    {"name": "Ghaziabad", "tier": "2", "lat": 28.6692, "lon": 77.4538},
    {"name": "Ludhiana", "tier": "2", "lat": 30.9010, "lon": 75.8573},
    {"name": "Agra", "tier": "2", "lat": 27.1767, "lon": 78.0081},
    {"name": "Nashik", "tier": "2", "lat": 19.9975, "lon": 73.7898},
    {"name": "Faridabad", "tier": "2", "lat": 28.4089, "lon": 77.3178},
    {"name": "Meerut", "tier": "2", "lat": 28.9845, "lon": 77.7064},
    {"name": "Rajkot", "tier": "2", "lat": 22.3039, "lon": 70.8022},
    {"name": "Varanasi", "tier": "2", "lat": 25.3176, "lon": 82.9739},
    {"name": "Srinagar", "tier": "2", "lat": 34.0837, "lon": 74.7973},
    {"name": "Aurangabad", "tier": "2", "lat": 19.8762, "lon": 75.3433},
    {"name": "Dhanbad", "tier": "2", "lat": 23.7957, "lon": 86.4304},
    {"name": "Amritsar", "tier": "2", "lat": 31.6340, "lon": 74.8723},
    {"name": "Allahabad", "tier": "2", "lat": 25.4358, "lon": 81.8463},
    {"name": "Ranchi", "tier": "2", "lat": 23.3441, "lon": 85.3096},
    {"name": "Coimbatore", "tier": "2", "lat": 11.0168, "lon": 76.9558},
    {"name": "Jabalpur", "tier": "2", "lat": 23.1815, "lon": 79.9864},
    {"name": "Gwalior", "tier": "2", "lat": 26.2183, "lon": 78.1828},
    {"name": "Vijayawada", "tier": "2", "lat": 16.5062, "lon": 80.6480},
    {"name": "Jodhpur", "tier": "2", "lat": 26.2389, "lon": 73.0243},
    {"name": "Madurai", "tier": "2", "lat": 9.9252, "lon": 78.1198},
    {"name": "Raipur", "tier": "2", "lat": 21.2514, "lon": 81.6296},
    {"name": "Kota", "tier": "2", "lat": 25.2138, "lon": 75.8648},
    {"name": "Chandigarh", "tier": "2", "lat": 30.7333, "lon": 76.7794},
    {"name": "Mysore", "tier": "2", "lat": 12.2958, "lon": 76.6394},
    
    # --- TIER 3 ---
    {"name": "Hubli", "tier": "3", "lat": 15.3647, "lon": 75.1240},
    {"name": "Jalandhar", "tier": "3", "lat": 31.3260, "lon": 75.5762},
    {"name": "Bhubaneswar", "tier": "3", "lat": 20.2961, "lon": 85.8245},
    {"name": "Salem", "tier": "3", "lat": 11.6643, "lon": 78.1460},
    {"name": "Tiruchirappalli", "tier": "3", "lat": 10.7905, "lon": 78.7047},
    {"name": "Bareilly", "tier": "3", "lat": 28.3670, "lon": 79.4304},
    {"name": "Aligarh", "tier": "3", "lat": 27.8974, "lon": 78.0880},
    {"name": "Tiruppur", "tier": "3", "lat": 11.1085, "lon": 77.3411},
    {"name": "Moradabad", "tier": "3", "lat": 28.8386, "lon": 78.7732},
    {"name": "Gurgaon", "tier": "3", "lat": 28.4595, "lon": 77.0266},
    {"name": "Jamshedpur", "tier": "3", "lat": 22.8046, "lon": 86.2029},
    {"name": "Dehradun", "tier": "3", "lat": 30.3165, "lon": 78.0322},
    {"name": "Udaipur", "tier": "3", "lat": 24.5854, "lon": 73.7125},
    {"name": "Ajmer", "tier": "3", "lat": 26.4499, "lon": 74.6399},
    {"name": "Ujjain", "tier": "3", "lat": 23.1765, "lon": 75.7811},
    {"name": "Jhansi", "tier": "3", "lat": 25.4484, "lon": 78.5685},
    {"name": "Jammu", "tier": "3", "lat": 32.7266, "lon": 74.8570},
    {"name": "Mangalore", "tier": "3", "lat": 12.9141, "lon": 74.8560},
    {"name": "Belgaum", "tier": "3", "lat": 15.8497, "lon": 74.4977},
    {"name": "Kochi", "tier": "3", "lat": 9.9312, "lon": 76.2673},
    {"name": "Noida", "tier": "3", "lat": 28.5355, "lon": 77.3910},
    {"name": "Rourkela", "tier": "3", "lat": 22.2604, "lon": 84.8536},
    {"name": "Nanded", "tier": "3", "lat": 19.1383, "lon": 77.3210},
    {"name": "Kolhapur", "tier": "3", "lat": 16.7050, "lon": 74.2433},
    {"name": "Akola", "tier": "3", "lat": 20.7002, "lon": 77.0082}
]

# --- DSA: THE MIN-HEAP ---
scheduler_heap = []

# --- HELPERS ---
def get_session():
    """Robust Session with Retries."""
    session = requests.Session()
    retry = Retry(total=3, backoff_factor=1, status_forcelist=[500, 502, 503, 504])
    adapter = HTTPAdapter(max_retries=retry)
    session.mount('http://', adapter)
    session.mount('https://', adapter)
    return session

def get_weather_description(code):
    if code == 0: return "Clear Sky"
    if code in [1, 2, 3]: return "Partly Cloudy"
    if code in [45, 48]: return "Foggy"
    if code in [51, 53, 55]: return "Drizzle"
    if code in [61, 63, 65]: return "Rain"
    if code in [71, 73, 75]: return "Snow"
    if code in [80, 81, 82]: return "Heavy Rain"
    if code in [95, 96, 99]: return "Thunderstorm"
    return "Unknown"

# --- DSA FUNCTIONS ---
def initialize_scheduler():
    """Populates Heap with (Time, Name, CityData) to prevent crashes."""
    now = time.time()
    for city in CITIES:
        # Pushing a tuple with Name as tie-breaker!
        heapq.heappush(scheduler_heap, (now, city['name'], city))
    print(f"🧠 Scheduler initialized with {len(scheduler_heap)} cities.")

def get_next_task():
    """Returns the top item from the heap."""
    if not scheduler_heap:
        return None, None, None
    return scheduler_heap[0] # Time, Name, Data

def reschedule_task(city_name, city_data):
    """Calculates next run time and pushes back to heap."""
    interval = TIER_INTERVALS.get(city_data['tier'], 3600)
    next_run = time.time() + interval
    
    heapq.heappop(scheduler_heap) # Remove current
    heapq.heappush(scheduler_heap, (next_run, city_name, city_data)) # Add future
    
    dt_str = datetime.fromtimestamp(next_run).strftime('%H:%M:%S')
    print(f"   ↪ Rescheduled {city_name} for {dt_str} ({int(interval/60)} mins)")

# --- PIPELINE FUNCTIONS ---
def setup_aqi_table():
    conn = None
    try:
        conn = mysql.connector.connect(host=AWS_HOST, user=AWS_USER, password=AWS_PASSWORD, database=AWS_DB_NAME, port=3306)
        cursor = conn.cursor()
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS city_aqi_tiers (
                id INT AUTO_INCREMENT PRIMARY KEY, city VARCHAR(100), tier VARCHAR(50), 
                aqi INT, temperature DECIMAL(5,2), humidity INT, 
                weather_condition VARCHAR(50), timestamp DATETIME
            )
        """)
        # Schema Migration Logic
        for col, dtype in [('temperature', 'DECIMAL(5,2)'), ('humidity', 'INT'), ('weather_condition', 'VARCHAR(50)'), ('aqi', 'INT')]:
            try: cursor.execute(f"ALTER TABLE city_aqi_tiers ADD COLUMN {col} {dtype}")
            except: pass
        conn.commit()
    except Exception as e:
        print(f"❌ DB Setup Error: {e}")
    finally:
        if conn and conn.is_connected(): conn.close()

def fetch_and_save(city):
    lat, lon = city['lat'], city['lon']
    session = get_session()
    w_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current=temperature_2m,relative_humidity_2m,weathercode&timezone=auto"
    a_url = f"https://air-quality-api.open-meteo.com/v1/air-quality?latitude={lat}&longitude={lon}&current=us_aqi"
    
    try:
        w_res = session.get(w_url, timeout=5)
        a_res = session.get(a_url, timeout=5)
        if w_res.status_code != 200 or a_res.status_code != 200: return False
        
        w = w_res.json().get('current', {})
        a = a_res.json().get('current', {})
        
        conn = mysql.connector.connect(host=AWS_HOST, user=AWS_USER, password=AWS_PASSWORD, database=AWS_DB_NAME, port=3306)
        cursor = conn.cursor()
        ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        
        q = "INSERT INTO city_aqi_tiers (city, tier, temperature, humidity, aqi, weather_condition, timestamp) VALUES (%s,%s,%s,%s,%s,%s,%s)"
        cursor.execute(q, (city['name'], city['tier'], w.get('temperature_2m'), w.get('relative_humidity_2m'), a.get('us_aqi'), get_weather_description(w.get('weathercode')), ts))
        conn.commit()
        conn.close()
        
        print(f"✅ {city['name']}: {w.get('temperature_2m')}°C, AQI {a.get('us_aqi')}")
        return True
    except Exception as e:
        print(f"❌ Error {city['name']}: {e}")
        return False

# --- MAIN ---
if __name__ == "__main__":
    print("🚀 Starting Smart DSA Scheduler (Priority Queue)...")
    setup_aqi_table()
    initialize_scheduler()
    
    try:
        while True:
            next_time, city_name, city_data = get_next_task()
            current_time = time.time()
            
            if current_time >= next_time:
                fetch_and_save(city_data)
                reschedule_task(city_name, city_data)
                time.sleep(0.1) # Prevent CPU spin
            else:
                wait = next_time - current_time
                if wait > 0:
                    print(f"💤 Sleeping for {int(wait)}s until next job...")
                    time.sleep(min(wait, 5))
            
    except KeyboardInterrupt:
        print("\n🛑 Scheduler Stopped.")
```
